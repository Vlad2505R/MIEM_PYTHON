## 📘 Формулировка задачи

Представьте, что вы работаете аналитиком в HR-отделе компании. Вам нужно смоделировать распределение зарплат сотрудников в трёх отделах (IT, Sales, HR) и построить визуализацию для сравнения.

### Используя numpy, сгенерируйте выборки зарплат:

**IT-отдел: средняя зарплата ≈ 120 000 руб., стандартное отклонение 20 000 руб.**

**Sales-отдел: средняя зарплата ≈ 80 000 руб., стандартное отклонение 15 000 руб.**

**HR-отдел: средняя зарплата ≈ 60 000 руб., стандартное отклонение 10 000 руб.**

**Каждая выборка — 200 сотрудников.**

### Постройте с помощью matplotlib:

**Гистограммы распределения зарплат для трёх отделов на одном графике.**

**Коробчатую диаграмму (boxplot) для сравнения.**

### Вычислите и выведите:

**среднее, медиану и стандартное отклонение для каждого отдела.**

**отдел с наибольшей вариативностью зарплат.**

### 🔹 Python (базовые вещи)

zip() — объединяет несколько списков/массивов в пары для перебора.

Циклы for, форматирование строк f"...".

Работа со списками и массивами.

### 🔹 NumPy (ключевые методы)

np.random.normal(mean, std, size) — генерация случайных данных (нормальное распределение).

np.mean(arr) — среднее значение.

np.median(arr) — медиана.

np.std(arr) — стандартное отклонение (разброс данных).

np.arange(start, stop, step) и np.linspace(start, stop, num) — создание массивов.

Индексация и срезы массивов (например, arr[0:10]).

### 🔹 Matplotlib (минимум)

plt.hist() — гистограмма.

plt.boxplot() — boxplot.

plt.plot() — линейный график.

plt.show() — показать график.

```python
# Подключаем библиотеки:
# numpy — для работы с массивами и генерацией случайных данных,
# matplotlib.pyplot — для построения графиков.
import numpy as np
import matplotlib.pyplot as plt

# Фиксируем "зерно" генератора случайных чисел,
# чтобы каждый запуск кода давал одинаковые результаты (удобно для обучения и отладки).
np.random.seed(42)

# Генерируем зарплаты в IT-отделе:
# распределение нормальное (колокол), среднее = 120 000, стандартное отклонение = 20 000,
# количество сотрудников = 200.
it_salaries = np.random.normal(120_000, 20_000, 200)

# Аналогично генерируем зарплаты в Sales-отделе: среднее 80 000, разброс 15 000.
sales_salaries = np.random.normal(80_000, 15_000, 200)

# Аналогично для HR-отдела: среднее 60 000, разброс 10 000.
hr_salaries = np.random.normal(60_000, 10_000, 200)

# ---------- ВИЗУАЛИЗАЦИЯ №1: ГИСТОГРАММЫ ----------

# Создаём окно графика с размером 10 на 6 дюймов.
plt.figure(figsize=(10, 6))

# Строим гистограмму зарплат в IT, разбиваем диапазон на 20 интервалов (bins),
# делаем прозрачность 0.6, добавляем подпись "IT".
plt.hist(it_salaries, bins=20, alpha=0.6, label="IT")

# То же для Sales.
plt.hist(sales_salaries, bins=20, alpha=0.6, label="Sales")

# То же для HR.
plt.hist(hr_salaries, bins=20, alpha=0.6, label="HR")

# Добавляем заголовок графика.
plt.title("Распределение зарплат сотрудников")

# Подписываем ось X — это зарплата в рублях.
plt.xlabel("Зарплата, руб.")

# Подписываем ось Y — количество сотрудников в каждом интервале зарплат.
plt.ylabel("Количество сотрудников")

# Добавляем легенду (чтобы различать отделы по цветам).
plt.legend()

# Показываем построенный график.
plt.show()

# ---------- ВИЗУАЛИЗАЦИЯ №2: BOXPLOT ----------

# Создаём окно графика размером 8 на 5 дюймов.
plt.figure(figsize=(8, 5))

# Строим boxplot (коробчатую диаграмму) для трёх отделов.
# Это позволяет сравнить медиану, квартили и выбросы.
plt.boxplot([it_salaries, sales_salaries, hr_salaries], labels=["IT", "Sales", "HR"])

# Добавляем заголовок.
plt.title("Сравнение зарплат по отделам (boxplot)")

# Подписываем ось Y.
plt.ylabel("Зарплата, руб.")

# Показываем график.
plt.show()

# ---------- РАСЧЁТ СТАТИСТИК ----------

# Перебираем пары "название отдела" и "массив данных"
for dept, data in zip(["IT", "Sales", "HR"], [it_salaries, sales_salaries, hr_salaries]):
    # Выводим среднее, медиану и стандартное отклонение (округляем до целого)
    print(f"{dept}: Среднее={np.mean(data):.0f}, "
          f"Медиана={np.median(data):.0f}, "
          f"Std={np.std(data):.0f}")
```

## 📝 Задача

Ситуация:
Вы работаете аналитиком в консалтинговой компании. Нужно смоделировать данные о ежемесячных расходах семей в трёх городах (Москва, Санкт-Петербург, Казань) и сравнить их.

### С помощью numpy смоделируйте расходы 300 семей в каждом городе.

**Москва: среднее ≈ 120 000 руб., стандартное отклонение 25 000.**

**Санкт-Петербург: среднее ≈ 90 000 руб., стандартное отклонение 20 000.**

**Казань: среднее ≈ 70 000 руб., стандартное отклонение 15 000.**

### Постройте с помощью matplotlib:

**Гистограммы распределения расходов в трёх городах (на одном графике).**

**Boxplot для сравнения.**

### Дополнительно рассчитайте для каждого города:

**среднее, медиану, стандартное отклонение.**

**процент семей, у которых расходы выше 100 000 руб.**

```python
# Подключаем библиотеки
import numpy as np
import matplotlib.pyplot as plt

# Фиксируем "зерно" для воспроизводимости результатов
np.random.seed(42)

# Генерируем данные о ежемесячных расходах (в рублях) для 300 семей
# Москва: среднее = 120 000 руб., std = 25 000
moscow_expenses = np.clip(np.random.normal(120_000, 25_000, 300), 0, None)

# Санкт-Петербург: среднее = 90 000 руб., std = 20 000
spb_expenses = np.clip(np.random.normal(90_000, 20_000, 300), 0, None)

# Казань: среднее = 70 000 руб., std = 15 000
kazan_expenses = np.clip(np.random.normal(70_000, 15_000, 300), 0, None)

# ---------- ВИЗУАЛИЗАЦИЯ №1: ГИСТОГРАММЫ ----------

# Создаём окно графика размером 10 на 6 дюймов
plt.figure(figsize=(10, 6))

# Гистограмма для Москвы
plt.hist(moscow_expenses, bins=20, alpha=0.6, label="Москва")

# Гистограмма для Санкт-Петербурга
plt.hist(spb_expenses, bins=20, alpha=0.6, label="Санкт-Петербург")

# Гистограмма для Казани
plt.hist(kazan_expenses, bins=20, alpha=0.6, label="Казань")

# Заголовок и подписи осей
plt.title("Распределение ежемесячных расходов семей по городам")
plt.xlabel("Расходы, руб.")
plt.ylabel("Количество семей")

# Добавляем легенду
plt.legend()

# Показываем график
plt.show()

# ---------- ВИЗУАЛИЗАЦИЯ №2: BOXPLOT ----------

# Создаём окно графика размером 8 на 5 дюймов
plt.figure(figsize=(8, 5))

# Boxplot для трёх городов
plt.boxplot([moscow_expenses, spb_expenses, kazan_expenses], 
            tick_labels=["Москва", "Санкт-Петербург", "Казань"])

# Заголовок и подписи осей
plt.title("Сравнение ежемесячных расходов по городам (boxplot)")
plt.ylabel("Расходы, руб.")

# Показываем график
plt.show()

# ---------- РАСЧЁТ СТАТИСТИК ----------

# Перебираем города и их данные
for city, data in zip(["Москва", "Санкт-Петербург", "Казань"], 
                      [moscow_expenses, spb_expenses, kazan_expenses]):
    # Рассчитываем среднее, медиану и стандартное отклонение
    mean = np.mean(data)
    median = np.median(data)
    std = np.std(data)
    # Рассчитываем процент семей с расходами выше 100 000 руб.
    percent_over_100k = (np.sum(data > 100_000) / len(data)) * 100
    
    # Выводим результаты, округленные до двух знаков
    print(f"{city}: Среднее={mean:.2f} руб., Медиана={median:.2f} руб., "
          f"Std={std:.2f} руб., Процент > 100 000 руб. = {percent_over_100k:.2f}%")
```

## 📝 Задача для самостоятельной работы

Ситуация:
Вы исследуете цифровое поведение студентов. Нужно смоделировать данные о том, сколько времени в среднем студенты проводят в социальных сетях в день, и сравнить факультеты.

### С помощью numpy сгенерируйте выборки для трёх факультетов (по 250 студентов в каждом):

Факультет информатики: среднее = 3.5 часа, std = 1.5.

Экономический факультет: среднее = 2.5 часа, std = 1.0.

Филологический факультет: среднее = 2.0 часа, std = 0.8.

### С помощью matplotlib постройте:

Гистограммы для трёх факультетов на одном графике.

Boxplot для сравнения.

### Рассчитайте:

среднее, медиану и стандартное отклонение для каждого факультета.

процент студентов, проводящих в соцсетях более 4 часов в день.

```python
# Подключаем библиотеки
import numpy as np
import matplotlib.pyplot as plt

# Фиксируем "зерно" для воспроизводимости результатов
np.random.seed(42)

# Генерируем данные о времени в соцсетях (в часах)
# Факультет информатики: среднее = 3.5 часа, std = 1.5, 250 студентов
it_faculty = np.random.normal(3.5, 1.5, 250)

# Экономический факультет: среднее = 2.5 часа, std = 1.0, 250 студентов
econ_faculty = np.random.normal(2.5, 1.0, 250)

# Филологический факультет: среднее = 2.0 часа, std = 0.8, 250 студентов
phil_faculty = np.random.normal(2.0, 0.8, 250)

# ---------- ВИЗУАЛИЗАЦИЯ №1: ГИСТОГРАММЫ ----------

# Создаём окно графика размером 10 на 6 дюймов
plt.figure(figsize=(10, 6))

# Гистограмма для факультета информатики
plt.hist(it_faculty, bins=20, alpha=0.6, label="Информатика")

# Гистограмма для экономического факультета
plt.hist(econ_faculty, bins=20, alpha=0.6, label="Экономика")

# Гистограмма для филологического факультета
plt.hist(phil_faculty, bins=20, alpha=0.6, label="Филология")

# Заголовок и подписи осей
plt.title("Распределение времени в соцсетях по факультетам")
plt.xlabel("Время в соцсетях, часы")
plt.ylabel("Количество студентов")

# Добавляем легенду
plt.legend()

# Показываем график
plt.show()

# ---------- ВИЗУАЛИЗАЦИЯ №2: BOXPLOT ----------

# Создаём окно графика размером 8 на 5 дюймов
plt.figure(figsize=(8, 5))

# Boxplot для трёх факультетов с использованием tick_labels
plt.boxplot([it_faculty, econ_faculty, phil_faculty], tick_labels=["Информатика", "Экономика", "Филология"])

# Заголовок и подписи осей
plt.title("Сравнение времени в соцсетях по факультетам (boxplot)")
plt.ylabel("Время в соцсетях, часы")

# Показываем график
plt.show()

# ---------- РАСЧЁТ СТАТИСТИК ----------

# Перебираем факультеты и их данные
for dept, data in zip(["Информатика", "Экономика", "Филология"], [it_faculty, econ_faculty, phil_faculty]):
    # Рассчитываем среднее, медиану и стандартное отклонение
    mean = np.mean(data)
    median = np.median(data)
    std = np.std(data)
    # Рассчитываем процент студентов, проводящих более 4 часов в соцсетях
    percent_over_4h = (np.sum(data > 4) / len(data)) * 100

    # Выводим результаты, округленные до двух знаков
    print(f"{dept}: Среднее={mean:.2f} ч, Медиана={median:.2f} ч, Std={std:.2f} ч, "
          f"Процент > 4 ч = {percent_over_4h:.2f}%")
```
